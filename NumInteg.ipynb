```python
from scipy.integrate import quad

# Function to integrate
def f(t):
    return t**2

# Integrate f(t) over [0, 1]
I, E = quad(f, 0, 1)

print("Result =", I)
print("Estimated error =", E)
```

**1. Rectangular Rules**

1.1. Midpoint rule

```python
import numpy as np

def f(t):
    return t**2

a = 0
b = 1
n = 1000  # number of rectangles

h = (b - a) / n
midpoints = a + h * (np.arange(n) + 0.5)  # midpoint of each interval
Jmp = h * np.sum(f(midpoints))

print("Midpoint rule approximation =", Jmp)
Emp = abs(I-Jmp)
Emp
```

1.2. Left endpoint rule

```python
t_left = np.linspace(a, b - h, n)
Jl = h * np.sum(f(t_left))
print("Left rule approximation =", Jl)
El = abs(I-Jl)
El
```

1.3. Right endpoint rule

```python
t_right = np.linspace(a + h, b, n)
Jr = h * np.sum(f(t_right))
print("Right rule approximation =", Jr)
Er = abs(I-Jl)
Er
```

**2. Trapezoidal Rule**

```python
import numpy as np

def f(t):
    return t**2

a = 0
b = 1
n = 1000  # number of subdivisions

t = np.linspace(a, b, n+1)
y = f(t)
Jt = (b - a) / n * (np.sum(y) - 0.5*y[0] - 0.5*y[-1])

print("Integral ≈", Jt)
Et = abs(I-Jt)
Et
```

**3. Simpson’s Rule**

```python
from scipy.integrate import simpson
import numpy as np

def f(t):
    return t**2

a = 0
b = 1
n = 1000  # must be even

t = np.linspace(a, b, n+1)
y = f(t)

Js = simpson(y, t)
print("Integral ≈", Js)
Es = abs(I-Js)
Es
```
