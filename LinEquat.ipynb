**1. Jacobi Method**

```python
import numpy as np

def jacobi(A, b, x0, tol=1e-6, max_iter=100):
    n = len(b)
    x = x0.copy()
    D = np.diag(A)
    R = A - np.diagflat(D)

    for k in range(max_iter):
        x_new = (b - np.dot(R, x)) / D
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k+1
        x = x_new
    raise Exception("Did not converge")

# Example system
A = np.array([[10, -1, 2],
              [-1, 11, -1],
              [2, -1, 10]], dtype=float)

b = np.array([6, 25, -11], dtype=float)
x0 = np.zeros_like(b)

x, iters = jacobi(A, b, x0)
print("Solution:", x)
print("Iterations:", iters)
```

**2. Gauss-Seidel Method**

```python
def gauss_seidel(A, b, x0, tol=1e-6, max_iter=100):
    n = len(b)
    x = x0.copy()

    for k in range(max_iter):
        x_new = x.copy()
        for i in range(n):
            s1 = sum(A[i][j] * x_new[j] for j in range(i))
            s2 = sum(A[i][j] * x[j] for j in range(i + 1, n))
            x_new[i] = (b[i] - s1 - s2) / A[i][i]
        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            return x_new, k+1
        x = x_new
    raise Exception("Did not converge")
##
# Example system
A = np.array([[10, -1, 2],
              [-1, 11, -1],
              [2, -1, 10]], dtype=float)

b = np.array([6, 25, -11], dtype=float)
x0 = np.zeros_like(b)

x, iters = jacobi(A, b, x0)
print("Solution:", x)
print("Iterations:", iters)
```
